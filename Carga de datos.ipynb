```python
import re

raw_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("sep", ";") \
    .option("inferSchema", "true") \
    .load("/databricks-datasets/wine-quality/winequality-red.csv")

clean_cols = [re.sub(r'[ ,;{}()\n\t=]+', '_', col.strip()) for col in raw_df.columns]
df_cleaned = raw_df.toDF(*clean_cols)

table_name = "default.wine_quality_raw"

df_cleaned.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(table_name)

print(f"Tabla Delta '{table_name}' creada exitosamente con columnas limpias.")
```
